## Preparar el entorno

```python
# Librerías necesarias
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import sys
print("Python:", sys.version.split()[0])
print("NumPy:", np.__version__)
```

## Subir una imagen

```python
# Se abre un selector del ordenador. Admitimos PNG/JPEG.
from google.colab import files
from io import BytesIO
subidos = files.upload()
nombre = next(iter(subidos))
imagen = Image.open(BytesIO(subidos[nombre])).convert("RGB")
I_rgb = np.asarray(imagen)
print("Archivo:", nombre, "dimensiones:", I_rgb.shape, "tipo:", I_rgb.dtype)
plt.imshow(I_rgb)
plt.title("Imagen original")
plt.axis("off")
plt.show()
```

## Inspeccionar y convertir

Convertimos la imagen cargada a escala de grises (`I_gray`) y creamos su versión normalizada (`I`) para las transformaciones posteriores.


```python
# Convertimos la imagen RGB subida desde el PC a escala de grises.
# I_gray es una matriz 2D con intensidades entre 0 y 255.
I_gray = np.asarray(imagen.convert("L"))
# Para negativo y gamma trabajaremos con valores entre 0 y 1.
I = I_gray.astype(np.float64) / 255.0
print("RGB:", I_rgb.shape, I_rgb.dtype)
print("Grises:", I_gray.shape, I_gray.dtype)
print("Normalizada:", I.shape, I.dtype, "rango:", I.min(), I.max())
plt.imshow(I_gray, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen en escala de grises")
plt.axis("off")
plt.show()
```

## Brillo y saturación

```python
# Aumentamos 80 unidades sobre 255, evitando overflow de uint8.
delta = 80
J = np.clip(I_gray.astype(np.float64)+delta,0,255).astype(np.uint8)
saturacion = 100*np.mean(J==255)
print(f"Píxeles saturados: {saturacion:.2f}%")
fig, ax = plt.subplots(1,2,figsize=(10,4))
for a,im,t in zip(ax,[I_gray,J],["Original","Brillo +80"]):
    a.imshow(im,cmap="gray",vmin=0,vmax=255)
    a.set_title(t);a.axis("off")
plt.show()
```

## Histograma

```python
fig, ax = plt.subplots(1,2,figsize=(10,3))
ax[0].hist(I_gray.ravel(), bins=256, range=(0,256))
ax[0].set_title("Original")
ax[1].hist(J.ravel(), bins=256, range=(0,256))
ax[1].set_title("Brillo +80")
for a in ax:
    a.set_xlabel("Intensidad");a.set_ylabel("N.º de píxeles")
plt.tight_layout();plt.show()
```

##Negativo y gamma

```python
negativo = 1-I
gamma_clara = I**0.4
gamma_oscura = I**2
fig,ax=plt.subplots(1,3,figsize=(11,3))
for a,im,t in zip(ax,[negativo,gamma_clara,gamma_oscura],
                  ["Negativo","Gamma 0.4","Gamma 2"]):
    a.imshow(im,cmap="gray",vmin=0,vmax=1)
    a.set_title(t);a.axis("off")
plt.show()
```
